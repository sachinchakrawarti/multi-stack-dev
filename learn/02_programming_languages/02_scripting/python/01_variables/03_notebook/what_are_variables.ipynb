# What Are Variables?

## Learning Objectives

- Understand variables
- Create variables
- Assign values
- Print variables
- Follow naming rules


## Theory

Variables are names that refer to values stored in memory. They allow us to store and reuse data in a program.

```python
print("Hello, Python!")
```

## Creating Variables

```python

name = "Sachin"
age = 25

print(name)
print(age)
```

## Assigning Values

```python

city = "Delhi"
country = "India"

print(city)
print(country)
```

## Printing Variables

```python

language = "Python"

print(language)
```

## Multiple Assignment

```python

x, y, z = 10, 20, 30

print(x)
print(y)
print(z)
```

## Variable Types

```python

name = "John"
age = 20
height = 5.8
student = True

print(type(name))
print(type(age))
print(type(height))
print(type(student))
```

## Dynamic Typing

```python

value = 100
print(value)

value = "Python"
print(value)
```

## Naming Rules

```python

first_name = "Alice"
student_age = 21
_marks = 95

print(first_name)
print(student_age)
print(_marks)
```

## Invalid Variable Names

```python

# 2name = "John"
# first-name = "John"
# class = "Python"

print("Invalid examples are commented.")
```

## Constants

```python

PI = 3.14159

print(PI)
```

## Best Practices

```python

student_name = "Rahul"
total_marks = 450
is_passed = True

print(student_name)
print(total_marks)
print(is_passed)
```

## Common Mistakes

```python

name = "Alice"

# Name = "Bob"

print(name)
```
