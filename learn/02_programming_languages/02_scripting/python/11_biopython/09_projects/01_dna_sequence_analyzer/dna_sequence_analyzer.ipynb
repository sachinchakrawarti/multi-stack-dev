# DNA Sequence Analyzer

Interactive analysis of DNA sequences using **Biopython**.

**Topics covered**
- Loading sequences from FASTA / GenBank
- Nucleotide composition and GC content
- Molecular weight
- Reverse complement, transcription, translation
- ORF (Open Reading Frame) detection
- Restriction enzyme site mapping
- Protein physicochemical properties

## 1. Setup

Import the analyzer module and helper functions.

```python
import os, sys
sys.path.append(os.getcwd())
from dna_analyzer import (
    DNAAnalyzer, write_sample_fasta,
    write_sample_genbank, SAMPLE_DNA,
)
print('Biopython project ready')
```

## 2. Create sample data

Write a synthetic DNA sequence to FASTA and GenBank format.

```python
os.makedirs('sample_data', exist_ok=True)
write_sample_fasta('sample_data/sample.fasta')
write_sample_genbank('sample_data/sample.gb')
print(open('sample_data/sample.fasta').read())
```

## 3. Load sequence from FASTA

```python
analyzer = DNAAnalyzer.from_fasta('sample_data/sample.fasta')
print('ID :', analyzer.record.id)
print('Seq:', analyzer.record.seq)
```

## 4. Basic statistics

Length, GC%, nucleotide counts, molecular weight.

```python
print('Length :', len(analyzer.record.seq))
print('GC %   :', analyzer.gc_content())
print('AT %   :', analyzer.at_content())
print('Counts :', analyzer.nucleotide_counts())
print('MW     :', analyzer.molecular_weight(), 'Da')
```

## 5. Sequence transformations

Reverse complement, transcription to mRNA, and translation.

```python
print('Rev comp:', analyzer.reverse_complement()[:60])
print('mRNA    :', analyzer.transcribe()[:60])
print('Protein :', analyzer.translate()[:60])
```

## 6. ORF detection

Find open reading frames on both strands in all 3 frames.

```python
for orf in analyzer.find_orfs(min_length=30):
    print(orf)
```

## 7. Restriction enzyme sites

```python
print(analyzer.restriction_sites())
```

## 8. Protein properties

Molecular weight, isoelectric point, aromaticity, instability index, GRAVY.

```python
import pprint
pprint.pprint(analyzer.protein_properties())
```

## 9. Full report

```python
report = analyzer.report(min_orf_len=30)
print(report)
```

## 10. Save report to disk

```python
with open('sample_data/report.txt', 'w') as f:
    f.write(report)
print('Saved sample_data/report.txt')
```
